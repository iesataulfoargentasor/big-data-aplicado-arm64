## Qué ejecuta cada cosa
- HDFS y MinIO almacenan archivos y objetos.
- HiveServer2 y Trino ejecutan SQL; el metastore guarda el catálogo compartido.
- En `hive.base.tabla`, `hive` es el catálogo de Trino: Trino no delega la consulta en HiveServer2.
- Python normal corre en Jupyter. El cuaderno 04 envía un trabajo real a YARN.
- Hive está configurado con Tez local dentro de HiveServer2. Sus mensajes de progreso no bastan para deducir ejecución distribuida.

Los helpers `hive_sql` y `trino_sql` aceptan **una sentencia cada vez** y devuelven una tabla de pandas. Cierran la conexión automáticamente. Puedes ver su código en `clientes.py`.
